# Aegis research workflow
Fixture results only validate the pipeline. Use acquired historical data for research; reserve a final test period.

```python
from aegis.research import synthetic_bars, walk_forward
from aegis.backtest import Backtester, FillModel
from aegis.strategies import Baseline
bars = synthetic_bars(days=4, symbols=("SPY",), seed=7)
result = Backtester(Baseline("orb"), FillModel.named("conservative")).run(bars, synthetic=True)
result["metrics"]
```

```python
evaluation = walk_forward(bars, "orb", {"opening_minutes": [5, 15]}, train_days=2, validation_days=1, test_days=1, embargo_days=0, synthetic=True)
evaluation["folds"]
```

The short fixture windows are for testing software. Production research needs longer windows, adequate trade counts, realistic costs, and point-in-time data. Log the hypothesis before inspecting the final test.
